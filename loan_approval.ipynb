```python
import pandas as pd
```

```python
data=pd.read_csv('loan_approval_dataset.csv')
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.columns=data.columns.str.strip()
```

```python
for i in data.columns:
    if data[i].dtype=='object':
        data[i]=data[i].str.strip()
    else:
        continue
```

```python
data['self_employed']=data['self_employed'].replace({'Yes':1,'No':0})
```

```python
data.head()
```

```python
x=data[['no_of_dependents','self_employed','income_annum','loan_amount','loan_term','cibil_score','residential_assets_value','commercial_assets_value','luxury_assets_value','bank_asset_value']]
```

```python
x.head()
```

```python
y=data['loan_status']
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.metrics import accuracy_score
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf2=DecisionTreeClassifier()
clf2.fit(x_train,y_train)
```

```python
accuracy_score(y_test,clf2.predict(x_test))*100
```

```python
accuracy_score(y_train,clf2.predict(x_train))*100
```
